#### Exercise 17.1

A globular cluster consists of 100,000 stars of the solar absolute magnitude. Calculate the total apparent magnitude of the cluster, if its distance is 10 kpc.

```python
import astropy.units as u
from numpy import log10

M_sun = 4.79 # from section 13.1 table, absolute visual magnitude
N = 100000
d = 10*u.kpc

M_cluster = -2.5*log10(N)+M_sun # equation 4.14
m_cluster = M_cluster+5*log10(d.to(u.pc).value)-5 # distance modulus, equation 4.13

print(f"The total apparent magnitude of the globular cluster is {m_cluster:.2f}.")
```
