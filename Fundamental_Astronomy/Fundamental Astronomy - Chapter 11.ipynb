#### Exercise 11.3 

The mass absorption coefficient of a neutrino is $\kappa$ = 10−21 m$^2$ kg$^{−1}$. Find the mean free path at the centre of the Sun.

```python
import astropy.units as u

kappa = 10**(-21)*u.m**2/u.kg
rho_centre = 1.6*10**5*u.kg/u.m**3 # found at section 13.1 table
d = 1/(kappa*rho_centre) # check example 11.7

print(f"The mean free path of a neutrino at the centre of the Sun is {d.to(u.km):.0e} ({d.to(u.AU):.0f}).")
```
