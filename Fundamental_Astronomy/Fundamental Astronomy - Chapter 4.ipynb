#### Exercise 4.2 

The absolute magnitude of a star in the Andromeda galaxy (distance 690 kpc) is *M* = 5. It explodes as a supernova, becoming one billion (10$^9$) times brighter. What is its apparent magnitude?

```python
import astropy.units as u
from sympy import symbols, Eq, log, solve

M_change, m_change = symbols('mag_change app_mag_change')

M = 5
L_change = 10**9
d = 690*u.kpc.to('pc') # distance modulus is given in parsecs (equations 4.12 and 4.13)

# equation 4.14 gives the relation between absolute magnitudes and luminosities
eq_4_14 = Eq(M_change - M, -2.5*log(L_change, 10))
sol_4_14 = solve((eq_4_14), (M_change))

# equation 4.13 gives the relation between the apparent magnitude, absolute magnitude, and distance in parsecs
eq_4_13 = Eq(m_change - sol_4_14[0], 5*log(d, 10)-5)
sol_4_13 = solve((eq_4_13), (m_change))

print(f"After explosion its apparent magnitude is {sol_4_13[0]:.2f}.")
```
