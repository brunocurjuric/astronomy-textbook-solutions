#### Exercise 19.1 

The galaxy NGC 772 is an Sb spiral, similar to M31. Its angular diameter is 7$^\prime$ and apparent magnitude 12.0. The corresponding values of M31 are 3.0$^◦$ and 5.0. Find the ratio of the distances of the galaxies

(a) assuming their sizes are equal,

(b) assuming they are equally bright.

```python
from astropy import units as u

angdim_ngc772 = 7*u.arcmin
m_ngc772 = 12.0
angdim_m31 = 3*u.degree
m_m31 = 5.0

# (a)
# if they have equal sizes, r is constant from size=distance*angular diameter, so their ratio of distances is just the ratio of angular diameters:
distance_ratio_size = (angdim_m31/angdim_ngc772).to('')
print(f"Assuming the sizes of galaxies are equal, the ratio of their distances is {distance_ratio_size:.2f}.")

# (b)
# if they are equally bright, their absolute magnitudes are the same, so from distance modulus m-M=5log(d)-5 we can find the ratios:
distance_ratio_brightness = 10**((m_ngc772-m_m31)/5)
print(f"Assuming the galaxies are equally bright, the ratio of their distances is {distance_ratio_brightness:.2f}.")
```
